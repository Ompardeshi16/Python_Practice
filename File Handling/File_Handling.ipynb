# File Handling

```python
FilePath = "File.txt"
file = open(FilePath, 'w')     # Create file File.txt
file.write("This is python class.")      # Write in the File.txt
file.write(" Hello Everyone " \
"It's File Handling Session")
file.close()
```

```python
FilePath = "File.txt"
file = open(FilePath, 'r')
file.read()
```

```python
FilePath = "File.txt"
file = open(FilePath, 'r')
newlines = file.readlines()
for i in newlines:  # Using Loop
    print(i)
```

```python
# Remove File
import os
FilePath = "File1.txt"
file = open(FilePath, 'w')
file.close()
os.remove(FilePath)
```

```python
# with keyworld -> Automaticaly close file after work done
FilePath = "File.txt"
with open(FilePath, 'w') as file:
    file.write("This is temporary file!")

with open(FilePath,'r') as file:
    print(file.read())
```
