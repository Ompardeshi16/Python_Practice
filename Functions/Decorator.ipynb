# <h1 style=color:red>Decorator</h1>

```python
# Method 1
def my_Func_Decor(func):
    def wrapper(x,y):
        if x>y:
           func(x,y)
        else:
            x,y = y,x
            print("Value Swapped")
            func(x,y)
    return wrapper 

@my_Func_Decor
def OriginalFun(x,y):
    print(x/y)
OriginalFun(5,10)
```

```python
# Method 2
def my_Func_Decor(func):
    def wrapper(x,y):
        if y>x:
            x,y = y,x
            print("Value Swapped")
            func(x,y)
    return wrapper 

@my_Func_Decor
def OriginalFun(x,y):
    print(x/y)
OriginalFun(5,10)
```

```python
# Convert values into int from string
def my_devi_decor(func):
    def wrapper(x,y):
        if isinstance(x, str):
            x = int(x)
        if isinstance(y, str):
            y = int(y)
            func(x,y)
        if y>x:
            x,y = y,x
            print("Value Swapped")
        return func(x,y)
    return wrapper

@my_devi_decor
def OriginalFun(x,y):
    print(x/y)
OriginalFun('5',10)
```
