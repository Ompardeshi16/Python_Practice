# Return and Non Return Functions

```python
def add_values(a,b):
    c = a+b
    return c # return excute's the code(store only single value)

value = add_values(3,2)
value # Return
```

```python
def sum_values(c,d):
    e = c+d
    return e

newValue = sum_values(value,5)
newValue
```

```python
def add_values(a,b):
    c= a+b
    print(c)

value = add_values(3,2)
print(value) # Non Return
```

- Yield Keywold(Generative Function)

```python
def sum_values():
    pass    # Empty Function

def add_values(a):
    for i in range(1,a):
        sum = i+i
        yield sum

value = add_values(5)

print(next(value))
print(next(value))
print(next(value))
print(next(value))
```
