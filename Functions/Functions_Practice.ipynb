- No Arguments

```python
def combineNames():
    FullName = "Om"+" "+"Pardeshi" 
    print(FullName)

combineNames()
```

- With Argument

```python
def printSquares(list):
    # NewDict = " "
    NewDict = {}
    for i in list:
        NewDict[i] = i*i
    print(NewDict)

list1 = [1,2,3,4,5]
printSquares(list1)
printSquares([6,7,8,9,10])
```

```python
list2= [11,12,13,14,15]
printSquares(list2)
```

2. Update the Dictionarie using Function

```python
my_dict = {"name": "Rajesh", "age": 24}
my_dict2 = {"pin":444001 ,"area": "Sanghavi"}

def updateDict(dict1, dict2):
    newDict = {**dict1,**dict2}
    print("Updated Dictionarie: ", newDict)

updateDict(my_dict, my_dict2)
```
