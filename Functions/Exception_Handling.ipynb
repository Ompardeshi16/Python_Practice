# Exception / Error Handling

```python
a=5
b=0
try:
    a/b
    # b/a   # Ignore except block due to no exception
except:
    print("ZeroDicisionError -> We can not divide a value by 0")
finally:
    print("Exception Handled!")
```

```python
a=5
b=0
try:
    c = a/b
except Exception as e: # Excute when error
    print(e, " -> We can not divide a value by 0")
finally: # Excute whenever error occur or not(At both)
    print("Code Excuted Successfully!")
```

```python
a=5
b=2
try:
    c = a/b
except Exception as e:
    print(e, " -> We can not divide a value by 0")
else: # Excute when no error
    print("No Error!")
```

```python
try:
    tup = (1,2,3,4)
    tup.append(5)
except Exception as e:
    print(e)
```

```python
newStr = "Ethan"
try:
    newInt = int(newStr)
except Exception as e:
    print("Exception: ", e)
finally:
    print("Detailed exception msg printed!")
```
