# String Functions in Python

- Case Change

Basic Functions

```python
# Upper
txt="hello friends"
x=txt.upper()
print(x)

# Lower
y=txt.lower()
print(y)

# Capitalize
z=txt.capitalize()
print(z)
```

```python
# Replace
txt="I like bananas"
x=txt.replace("bananas","apples")
print(x)

# Join
my_list = ["Apple", "Banana", "Grapes", "Mango"]
y = ",".join(my_list)
print(y)

# Split
txt = "I like to play football"
txt2 = "Welcome,to,the,world,of,programming"
z = txt.split(" ")
z2 = txt2.split(",")
print(z)
print(z2)
```

Search/Count Functions

```python
# Count
txt = "Hey! Welcome to the world of programming."
x = txt.count("o")
print("Count of 'o':", x)

# Find
y = txt.find("Welcome")
print("Index of 'Welcome':", y)

# index
z = txt.index("programming")
print("Index of 'programming':", z)

# startswith
a = txt.startswith(" Hey!")
print("Starts with ' Hey!':", a)

# endswith
b = txt.endswith("programming.")
print("Ends with 'programming.':", b)
```

String Validation

```python
# isnumeric
num = "12345"
p = num.isnumeric()
print("Is numeric:", p)
t = alpha.isnumeric()
print("Is numeric (alpha):", t)

# isdigit
digit = "67890"
q = digit.isdigit()
print("Is digit:", q)

# isalpha
alpha = "Hello"
r = alpha.isalpha()
print("Is alpha:", r)
u = num.isalpha()
print("Is alpha (num):", u)

# isalnum
alnum = "Hello123"
s = alnum.isalnum()
print("Is alphanumeric:", s)
```

```python
# Extracting Substrings
new_txt = "rajeshs40"
substring = new_txt[0:6]  # Extract characters from index 0 to 5
print("Extracted substring:", substring)

new_txt2 = "rajesh_patil40"
new_str = ""
for i in new_txt2:
    if i.isalpha() == True:
        new_str = new_str + i
print("Extracted substring from alpha numeric string:", new_str) # r ra raj raje rajes rajesh rajeshp rajeshpa rajeshpat rajeshpati rajeshpatil

new_str2 = ""
for i in new_txt2:
    if i.isnumeric() == True:
        new_str2 = new_str2 + i
print("Extracted substring(numeric) from alpha nymeric string:", new_str2) # 4 40
```

- Type Conversion

```python
# Know the datatypes
my_val_1 = "20"
my_val_2 = 20
my_val_3 = 20.00
print(type(my_val_1))
print(type(my_val_2))
print(type(my_val_3))
```

```python
# Type Conversion
print(int(my_val_1))
print(type(my_val_1))

print(float(my_val_2))
print(type(my_val_2))

print(str(my_val_3))
print(type(my_val_3))
```
