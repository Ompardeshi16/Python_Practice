# Tuple
- Access Items

```python
myTuple = ("apple", "banana", "cherry")
print(myTuple[1])
print(myTuple[-1])
```

- Slicing

```python
print(myTuple[:2])
print(myTuple[1:3])
print(myTuple[:])
```

- Search / Count

```python
print(myTuple.index("banana"))
myTuple.count("cherry")
```

- Add / Join

```python
myTuple2 = ("orange", "kiwi", "melon")
myTuple3 = myTuple + myTuple2
print(myTuple3)

myTuple += ("grapes", "pear")
print(myTuple)
```

- Repeat / Check Item

```python
print(("A", "B") * 3)

"kiwi" in myTuple2
```

- Length / Convert / Copy / Sorted

```python
print(len(myTuple))

print(tuple(["apple", "banana", "mango"]))

newTuple = myTuple2
print(newTuple)

print(sorted(myTuple2)) # Sort and return new List
```
