# List
- Slicing / Accessing

```python
mylist = ["apple", "banana", "cherry", "date", "pineapple"]
print(mylist[0])
print(mylist[-1])
```

- Add / Insert / Join / Merge

```python
mylist.append("kiwi") # Add element in last
print(mylist)

mylist.insert(1, "orange") # Add element at specified index
print(mylist)

mylist01 = ["kiwi", "orange"]
listjoined = mylist + mylist01 # Join lists
print(listjoined)

mylist.extend(["mango", "grape"]) # Add multiple elements in previous list
print(mylist)
```

- Remove

```python
mylist1 = ["apple", "banana", "cherry", "date", "pineapple"]
mylist1.remove("pineapple") # Remove specific element
print(mylist1)

mylist1.pop(2) # Remove element at specified index
print(mylist1)

mylist1.pop() # Remove last element by default
print(mylist1)

del mylist1[0] # Delete element at specific index
print(mylist1)
```

- Delete / Empty

```python
mylist1.clear() # Only clear the list
print(mylist1)

del mylist1
# print(mylist1)  # This will raise an error since mylist1 is deleted
```

- Copy

```python
l2 = mylist # Similar memory location
print(l2)

new_list = mylist.copy() # Different memory location
print(new_list)
```

- Sort

```python
l1 = [45,18,7,23,89,12]
l1.sort() # Sort the list in ascending order(Similar memory location)
print(l1)

new_l1 = sorted(l1) # Sort and return new list(Different memory location)
print(new_l1)

l1.sort(reverse=True) # Sort the list in descending order
print(l1)
```

- Reverse

```python
l1.reverse() # Reverse the list
print(l1)

l1[::-1] # Reverse using Slicing
```

- Search / Count

```python
print(mylist.index("cherry")) # Get index of specific item

print(mylist.count("banana")) # Count how many times it occurs
```

- Convert / Length

```python
print(list("ABCDEF")) # Convert string to list

len(mylist) # Get length of list
```

- List Comprehension

```python
# Creating new list of squares from first list

list1 = [1,2,3,4,5,6,7]
new_list = []
for i in list1:
    new_list.append(i*i)
print(new_list)
print()

# List Comprehension
print("List Comprehension:")
print([i*i for i in list1 if i<=5])
```
