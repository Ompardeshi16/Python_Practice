# Dictionary 
- Access Items

```python
my_dict = {"name": "Ethan", "age": 20, "city": "New York"}

print(my_dict["name"])
print(my_dict.get("age"))
```

- Add / Update Items

```python
my_dict["class"] = "BCA"
print(my_dict)

my_dict["age"] = 25
print(my_dict)

my_dict.update({"age": 30 , "city": "Jalgaon"})
```

```python
for i in my_dict:
    print(my_dict[i])
```

```python
for i in my_dict.keys():
    print(my_dict.get(i))
```

```python
for i,j in my_dict.items():
    print(i,j)
```

- Remove Items

```python
my_dict = {"name": "Ethan", "age": 20, "city": "New York"}
my_dict.pop("age")
print(my_dict)

my_dict.popitem() # Remove last inserted item
print(my_dict)

del my_dict["name"]
print(my_dict)

my_dict.clear()

del my_dict
```

- Keys / Values / Items

```python
my_dict = {"name": "Ethan", "age": 20, "city": "New York"}

print(my_dict.keys())

print(my_dict.values())

print(my_dict.items())
```

- Check Item / Length / Copy

```python
my_dict = {"name": "Ethan", "age": 20, "city": "New York"}

print("age" in my_dict)

print("Length:", len(my_dict))

new_dict = my_dict.copy()
print(new_dict)
```

- Create Dictionary / Merge / Default Value

```python
my_dict = {"name": "Ethan", "age": 20, "city": "New York"}
my_dict2 = {"area": "San francisco"}

new_dict = {**my_dict, **my_dict2}
print(new_dict)
new_dict2 = dict(my_dict, **my_dict2)
print(new_dict2)

my_dict.setdefault("country", "India")
```

- Reverse / Sorting

```python
my_dict = {"name": "Ethan", "age": 20, "city": "New York"}

new_dict = dict(reversed(list(my_dict.items())))
print(new_dict)

sort_dict = dict(sorted(my_dict.items()))
print(sort_dict)
```

- Dictionaries Comprehension

```python
list = [1,2,3]
newDict = {}

for i in list:
    newDict[i] = i**2

print(newDict)

# Comprehension
# {key : value Loop}
print("Dictionaries Comprehension:", {i : i**2 for i in list})
# {i : i*2 for i in list}
```
