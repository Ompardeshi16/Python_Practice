> # <h1 style="color: red">List Comprehension</h1>

```python
for i in range(1,10):
    print(i, end=" ")
```

```python
[i for i in range(1,10)]
```

```python
# [exp Loop if_condition]
# [exp for i in iterable if_condition]
# Print Even Numbers

[i for i in range(1,10)]
```

**Print Square of Even Numbers**

```python
[i**2 for i in range(1,10) if i%2==0]
```

**Multiple Conditions in List Comprehension**

```python
newList = []
for i in range(0,10):
    if i<2 or i>5:
        newList.append(i)
newList
```

**Using or operator**

```python
[i for i in range(0,10) if i<2 or i>5]
```

```python
newList2 = []
for i in range(0,10):
    if i%2==0 and i>5:
        newList2.append(i)
newList2
```

**Using and operator**

```python
[i for i in range(0,10) if i%2==0 and i>5]
```
