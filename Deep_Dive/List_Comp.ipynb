# <h1 style="color: red">List Comprehension</h1>

```python
for i in range(1,10):
    print(i, end=" ")
```

```python
# [exp Loop if_condition]
# [exp for i in iterable if_condition]
# Print Even Numbers

[i for i in range(1,10)]
```

**Print Square of Even Numbers**

```python
[i**2 for i in range(1,10) if i%2==0]
```

**Multiple Conditions in List Comprehension**

```python
newList = []
for i in range(0,10):
    if i<2 or i>5:
        newList.append(i)
newList
```

**Using or operator**

```python
[i for i in range(0,10) if i<2 or i>5]
```

```python
newList2 = []
for i in range(0,10):
    if i%2==0 and i>5:
        newList2.append(i)
newList2
```

**Using and operator**

```python
newList2 = []
for i in range(0,20):
    if i%2==0 and i>5:
        newList2.append(i)
newList2
```

```python
[i for i in range(0,10) if i%2==0 and i>5]
```

**Multiple Expresions(Output)**

```python
newList = []
for i in range(15):
    if i<5:
        newList.append(i*2)
    elif i<10:
        newList.append(i/2)
    else:
        newList.append(i+2)
newList
```

```python
# [i for i in range(15)]
[i*2 if i<5 else i/2 if i<10 else i+2 for i in range(15)]
```

```python
# Practice Question
# Convert in Dictionary comprehension
{i : i*2 for i in range(15)}
```

```python
# {exp1: exp2 Loop}
{i : i*2 if i<5 else i/2 if i<10 else i+2 for i in range(15)}   # {Key : Value}
```

```python
# Can add condition on either Key or Value
{i*2 if i<5 else i/2 if i<10 else i+2 : i for i in range(15)}
```
