> # <h1 style="color:red">**Instance Class Method**</h1>

> **Instant Variable and Instant Method**

```python
class Dog():
    def __init__(self):
        self.dage=14

dog1=Dog()
dog1.dage
```

```python
class Dog():
    def __init__(self,dage):
        self.dage = dage    # Instance Variable

    def abc(self):      # Instance Method
        print(self.dage)

    def my_dog_details(self,colour):    # Instance Method with argument
        print(f"My dog has {colour} colour")

dog1=Dog(14)
dog1.abc()
dog1.my_dog_details("brown")
```

> **Class Variables and Class Methods**

```python
class School:
    schoolName = "DNVK"    # Class Variable
School.schoolName
```

```python
sobj = School()
sobj.schoolName = "SSBT COET"
School.schoolName
```

**Banking Example**

```python
class Banking:
    Int_Rate = 0.08     # Interset Rate = 8%

    @classmethod
    def change_int_rate(cls):
        cls.Int_Rate = cls.Int_Rate + (cls.Int_Rate * 0.02)
        print(cls.Int_Rate)

print(f"Interest rate before: {Banking.Int_Rate}")
Banking.change_int_rate()
```
