> # <h1 style="color:red">**Instance Class Method**</h1>

```python
class Dog():
    def __init__(self):
        self.dage=14

dog1=Dog()
dog1.dage
```

```python
class Dog():
    def __init__(self,dage):
        self.dage = dage    # Instance Variable

    def abc(self):      # Instance Method
        print(self.dage)

    def my_dog_details(self,colour):
        print(f"My dog has {colour} colour")

dog1=Dog(14)
dog1.abc()
dog1.my_dog_details("brown")
```
