```python
# Find the matched values with the index Number
col_1 = ["Red", "Yellow", "LightGreen"]
col_2 = ["Black", "White", "LightGreen"]
col_3 = ["Black", "LightGreen", "White", "Pink"]

for i in col_1:
    for j in col_2:
        for k in col_3:
            if i==j==k:
                # print("Matched", i,j,k)
                print("Comparision in beteen both three columns:")
                indexList = [col_1.index(i), col_2.index(j), col_3.index(k)]
                print("Matched in both columns: ", i,j,k,"with index's ", indexList, "respectively.")
                print(" ")

```

```python
for j in col_2:
    for k in col_3:
        if j==k:
            indexList2 = [col_2.index(j), col_3.index(k)]
            print("Matched in column 2 and 3: ", j,k, "with index's ", indexList2, "respectively.")
```
