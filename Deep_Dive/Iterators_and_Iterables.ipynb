# <h1 style="color:red">Iterators and iterable</h1> 

```python
fruitsList = ["apple", "banana", "cherry", "orange"]
print(dir(fruitsList))
print()

for i in fruitsList:
    print(i)
```

**Apply Filter**

```python
fruitsList = [1,2,3,4,5,6,7,8]
def extractOnlyEvenNum(i):
    return i%2==0

abc = filter(extractOnlyEvenNum, fruitsList)
print(abc)
print(dir(abc))

for i in abc:
    print(i)
```

**Conversion to iter**

```python
mystr ="apple"
myit = iter(mystr)
print(myit)
print("")
print(dir(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
```

```python
fruitTup=("apple", "banana", "cherry")
myit = iter(fruitTup)
print(myit)
print("")
print(dir(myit))
print(next(myit))
print(next(myit))
print(next(myit))
```
