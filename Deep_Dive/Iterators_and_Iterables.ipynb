# <h1 style="color:red">Iterators and iterable</h1> 

```python
fruitsList = ["apple", "banana", "cherry", "orange"]
print(dir(fruitsList))
print()

for i in fruitsList:
    print(i)
```

**Apply Filter**

```python
fruitsList = [1,2,3,4,5,6,7,8]
def extractOnlyEvenNum(i):
    return i%2==0

abc = filter(extractOnlyEvenNum, fruitsList)
print(abc)
print(dir(abc))

for i in abc:
    print(i)
```
