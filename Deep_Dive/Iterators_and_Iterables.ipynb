# <h1 style="color:red">Iterators and iterable</h1> 

```python
fruitsList = ["apple", "banana", "cherry", "orange"]
print(dir(fruitsList))
print()

for i in fruitsList:
    print(i)
```

**Apply Filter**

```python
fruitsList = [1,2,3,4,5,6,7,8]
def extractOnlyEvenNum(i):
    return i%2==0

abc = filter(extractOnlyEvenNum, fruitsList)
print(abc)
print(dir(abc))

for i in abc:
    print(i)
```

**Conversion to iterator**

```python
mystr ="apple"  # Iterator(Like list) -> __iter__
myit = iter(mystr)  #Iterable -> __iter__ and __next__
print(myit)
print("")
print(dir(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
```

```python
fruitTup=("apple", "banana", "cherry")
myit = iter(fruitTup)
print(myit)
print("")
print(dir(myit))
print(next(myit))
print(next(myit))
print(next(myit))
```

```python
# Can apply loop only one time in case of iterators
newRange = range(1,10)
def new_func(i):
    if i%2==0:
        return i

new_iterator = filter(new_func, newRange)
print(new_iterator)

for i in new_iterator:
    print(i, end=" ")
```

```python
def new_func(i):
    if i%2==0:
        return i

new_iterator = filter(new_func, newRange)
print(new_iterator)
print(list(new_iterator))
```
