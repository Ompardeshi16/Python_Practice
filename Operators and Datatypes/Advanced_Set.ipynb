# Set
- Add Items / Remove Items

```python
mySet = {"apple", "banana", "mango"}
print(mySet)

mySet.add("kiwi")
print(mySet)
mySet.update(["orange", "grapes"])
print(mySet)

mySet.remove("banana")
print(mySet)
mySet.discard("banana") # No error if not present
print(mySet)
mySet.pop() 
print(mySet)
mySet.clear()
```

- Set Operations

```python
set1 = {1,2,3}
set2 = {3,4,5}

set3 = set1.union(set2)
print("Union:", set3)

set4 = set1.intersection(set2)
print("Intersection:", set4)

set5 = set1.difference(set2)
print("Difference:", set5)

set6 = set1.symmetric_difference(set2)
print("Symmetric Difference:", set6)

newSet = set1.copy()
print("Copy:", newSet)

set1.update(set2)
print("Updated Set1:", set1)
```
