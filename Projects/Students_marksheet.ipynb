```python
percentage = float(input("Enter your percentage (0–100): "))

if percentage < 0 or percentage > 100:
    print("Invalid percentage.")
elif percentage >= 90:
    print("Grade: A\nResult: Pass")
elif percentage >= 80:
    print("Grade: B\nResult: Pass")
elif percentage >= 70:
    print("Grade: C\nResult: Pass")
elif percentage >= 60:
    print("Grade: D\nResult: Pass")
elif percentage >= 40:
    print("Grade: E\nResult: Pass")
else:
    print("Grade: F\nResult: Failed")
```
