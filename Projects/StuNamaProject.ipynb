# Basic sample project to check fruits avaliable or not in fruit stall
- Get string (Fruits) as an input from user
- Use for loop for storing the fruits in list
- Check this fruits are available in fruits stall or not by using for loop and if else 
- Based on result print answer

```python
student_counts = int(input("Enter counts of students:"))
student_names = {"Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Grace", "Henry", "Ivy", "Jack"}
students_present = []

for i in range(student_counts):
    student = input("Enter student name:")
    students_present.append(student) # append -> adds the student to the list
print("Students present:", students_present)

for student in students_present:
    if student in student_names:
        print(student, "is in the attendance list")
    else:
        print(student, "is not in the attendance list")
```
