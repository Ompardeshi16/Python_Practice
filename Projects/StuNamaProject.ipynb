# Basic sample project to check fruits avaliable or not in fruit stall
- Get string (Fruits) as an input from user
- Use for loop for storing the fruits in list
- Check this fruits are available in fruits stall or not by using for loop and if else 
- Based on result print answer

```python
fruits_counts = int(input("Enter counts of fruits"))
fruits_stall = {"apple", "banana", "orange", "grapes", "mango", "kiwi", "papaya", "pear", "peach", "plum"}
fruits = []

for i in range(fruits_counts):
    fruit = input("Enter fruits name:")
    fruits.append(fruit) # append -> adds the fruit to the list
print("Customer wants:", fruits)

for fruit in fruits:
    if fruit in fruits_stall:
        print(fruit, "is available in the stall")
    else:
        print(fruit, "is not available in the stall")
```
