# <h1 style=color:red>Access Modifier</h1>

```python
# Inside the Class
class Student:
    def __init__(self):
        self.name = "Rahul"     # Public
        self._age = 20      # Protected
        self.__marks = 90       # Private

    def display(self):
        print(self.name)    # Public
        print(self._age)    # Protected
        print(self.__marks) # Private 

obj = Student()
obj.display()
```

```python
# Outside the Class
class Student:
    def __init__(self):
        self.name = "Rahul"     # Public
        self._age = 20      # Protected
        self.__marks = 90       # Private

obj = Student()
print(obj.name)    # Public
print(obj._age)    # Protected
print(obj.__marks) # Private - Error 
```

```python
# Inside child class Inheritance
class ParentClass:
    def __init__(self):
        self.name = "Rahul"     # Public
        self._age = 20      # Protected
        self.__marks = 90       # Private

class ChildClass(ParentClass):
    def display(self):
        print(self.name)    # Public - Allowed
        print(self._age)    # Protected - Allowed
        print(self.__marks) # Private  - Error

obj = ChildClass()
obj.display()
```
