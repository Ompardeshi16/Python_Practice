# Method Overloading

```python
class myClass:
    def sum(self, a=None, b=None, c=None):
        if a != None and b != None and c != None:
            return a+b+c
        elif a != None and b != None:
            return a+b
        else:
            return "Provide atleast 2 argument"

Obj = myClass()
sum = Obj.sum(5,10)
print(sum)
```
