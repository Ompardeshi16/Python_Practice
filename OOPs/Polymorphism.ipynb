# <h1 style=color:red>Method Overloading</h1>

```python
# Simple method without overloading
class myClass:
    def sum_4(self,a,b,c,d):
        return a+b+c+d
    def sum_3(self,a,b,c):
        return a+b+c
    def sum_2(self,a,b):
        return a+b

myObj = myClass()
print(myObj.sum_2(1,2))
print(myObj.sum_3(1,2,3))
print(myObj.sum_4(1,2,3,4))
```

```python
# Method Overloading
class myClass:
    def sum(self, a=None, b=None, c=None, d=None):
        if a != None and b != None and c != None and d != None:
            return a+b+c+d
        elif a != None and b != None and c != None:
            return a+b+c
        elif a != None and b != None:
            return a+b
        else:
            return "Provide atleast 2 argument"

Obj = myClass()
sum = Obj.sum(5,10,1,6)
print(sum)
```

# <h1 style=color:red>Method Overriding</h1>

```python
class Animal:
    def sound(self):
        return "Animal makes Sound!"

class Dog(Animal):
    def sound(self):
        return "Dog Barks!"

class Cat(Animal):
    def sound(self):
        return "Cat Meow!"

AniObj = Animal()
DogObj = Dog()
CatObj = Cat()

print(AniObj.sound())
print(DogObj.sound())
print(CatObj.sound())
```

```python
class P_C_Intersest:
    def Interest(self):
        return 0.011

class C_C_Interest(P_C_Intersest):
    def Interest(self):
        return 0.08

Home_Obj = C_C_Interest()
Home_Obj.Interest()
```
