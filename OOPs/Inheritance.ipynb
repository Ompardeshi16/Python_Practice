# <h1 style="color:red">Inheritance</h1>

```python
class A:    # Parent Class
    def method_class_A(self):
        return "This is parent Class Method"

class B:    # Child Class
    def method_class_B(self):
        return "This is child Class Method"

p_Obj_A = A()
print(p_Obj_A.method_class_A())
```

```python
c_Obj_B = B()
print(c_Obj_B.method_class_B())
```

```python
class A:    # Parent Class
    def method_class_A(self):
        return "This is parent Class Method"

class B(A):    # Child Class
    def method_class_B(self):
        return "This is child Class Method"

p_Obj_A = A()
print(p_Obj_A.method_class_A())

c_Obj_B = B()
print(c_Obj_B.method_class_B())
print("Calling parent method through child object: ", c_Obj_B.method_class_A())
```
