# <h1 style="color:red">Inheritance</h1>

```python
class A:    # Parent Class
    def method_class_A(self):
        return "This is parent Class Method"

class B:    # Child Class
    def method_class_B(self):
        return "This is child Class Method"

p_Obj_A = A()
print(p_Obj_A.method_class_A())
```

```python
class Parent:    # Parent Class
    def method_class_Parent(self):
        return "This is parent Class Method"

class Child:    # Child Class
    def method_class_Child(self):
        return "This is child Class Method"

p_Obj = Parent()
print(p_Obj.method_class_Parent())
c_Obj = Child()
print(c_Obj.method_class_Child())
```

```python
c_Obj_B = B()
print(c_Obj_B.method_class_B())
```

**Single Inheritance**

```python
class A:    # Parent Class
    def method_class_A(self):
        return "This is parent Class Method"

class B(A):    # Child Class
    def method_class_B(self):
        return "This is child Class Method"

p_Obj_A = A()
print(p_Obj_A.method_class_A())

c_Obj_B = B()
print(c_Obj_B.method_class_B())
print("Calling parent method through child object: ", c_Obj_B.method_class_A())
```

**MultiLine inheritance**

```python
class GrandParentClass:
    def GP_Method(self):
        return "GrandParent Method -> Have 5 Acres of Land"

class ParentClass(GrandParentClass):
    def P_Method(self):
        return "Parent Method -> Have 4BHK House"

class ChildClass(ParentClass):
    def C_method(self):
        return "Child Method -> Learning different skills"

childOBJ = ChildClass()
print(childOBJ.GP_Method())
print(childOBJ.P_Method())
print(childOBJ.C_method())
```

**Multiple Inheritance**

```python
class MotherClass:
    def M_Method(self):
        return "Mother Method -> Have Gold"

class FatherClass:
    def F_Method(self):
        return "Father Method -> Have 4BHK House"

class ChildClass(MotherClass, FatherClass):
    def C_method(self):
        return "Child Method -> Learning different skills"

childOBJ = ChildClass()
print(childOBJ.M_Method())
print(childOBJ.F_Method())
print(childOBJ.C_method())
```

```python
class FatherClass:
    def F_Method(self):
        return "Father Method -> Have 4BHK House"

class ChildClass1(FatherClass):
    def C_method1(self):
        return "Child Method 1 -> Learning different skills"

class ChildClass2(FatherClass):
    def C_method2(self):
        return "Child Method 2 -> Trying to pass Govn Exam"

childOBJ_1 = ChildClass1()
print(childOBJ_1.F_Method())
print(childOBJ_1.C_method1())
print("")

childOBJ_2 = ChildClass2()
print(childOBJ_2.F_Method())
print(childOBJ_2.C_method2())
```

**Practice que 1 -> Single Inheritance**

```python
class KeypadPhone:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def call(self, number, name):
        return f"{self.brand} {self.model}: Calling {name} -> {number}..."

    def send_sms(self, number, text):
        return f"SMS send to {number}: '{text}'"

class MobilePhone(KeypadPhone):
    def __init__(self, brand, model, battery_MAH):
       self.brand = brand
       self.model = model
       self.battery_MAH = battery_MAH

    def play_Game(self):
        return "Playing BGMI!"

    def listen_Music(self):
        return "Listening spotify music!" 

phoneOBJ = MobilePhone("Xiomi", "Note 9 pro max", 5000)
print(phoneOBJ.call("9022608319", "Mohit"))
print(phoneOBJ.send_sms("902608319", "Hey! is tommorow holiday ?"))
print(phoneOBJ.play_Game())
print(phoneOBJ.listen_Music())
```

**Practice que 2 -> MultiLevel Inheritance**

```python
class KeypadPhone:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def call(self, number, name):
        return f"{self.brand} {self.model}: Calling {name} -> {number}..."

    def send_sms(self, number, text):
        return f"SMS send to {number}: '{text}'"

class MobilePhone(KeypadPhone):
    def __init__(self, brand, model, battery_MAH):
       self.brand = brand
       self.model = model
       self.battery_MAH = battery_MAH

    def play_Game(self):
        return "Playing BGMI!"

    def listen_Music(self):
        return "Listening spotify music!" 

class SmartPhone(MobilePhone):
    def __init__(self, brand, model, battery_MAH, os_name):
       self.brand = brand
       self.model = model
       self.battery_MAH = battery_MAH
       self.os_name = os_name

    def browse_internet(self. site):
        return f"Opening {site} on {self.os_name}"

    def install_app(self, app):
        return f"Installing {app} on {self.os_name}"

    phone_obj = SmartPhone("Samsung", "Galaxy S24", 5500, "Android")
    print(phone_obj.call("9876543210"))
    print(phone_obj.play_game())
    print(phone_obj.browse_internet("google.com"))
    print(phone_obj.install_app)
```
