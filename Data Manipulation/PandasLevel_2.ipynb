**Use of Map**

```python
def newAirFare(x):
  return int(x) + 500

df['NewFare'] = df['Fare'].map(newAirFare)
df.head()
```

```python
 # Can use return int(x) * 1.10 instead of return int(x)+(x * 10) / 100
def newAirFare(x):

  return int(x)+(x * 10) / 100

df['NewFarePer'] = df['Fare'].map(newAirFare)
df.head()
```

**Working with index:**<br>
The index in a DataFrame is crucial for data alignment and access. Here are some common operations for working with indices:

1. Set Index (set column as Index)
df.set_index('column_name', inplace=True)
Replace 'column_name' with the name of the column to be set as the index.

2. Reindex (Reindex rows or columns)
df.reindex(new_index) for rows
df.reindex(columns=new_columns) for columns.
Provide new_index or new_columns with the new index or column labels.

3. Reset Index (Reset the index to default)
df.reset_index(drop=False, inplace=True)
Use drop=True to remove the index column from the DataFrame.

```python
# Set index
'''Make a column of dataframe as index with remove default index.'''
newDict = {'Date'           :['10/2/2011', '11/2/2011', '12/2/2011', '13/2/2011'],
            'Product'       :[' UMbreLla', '  maTtress', 'BaDmintoN ', 'Shuttle'],
            'Price'         :[1250, 1450, 1550, 400],
            'Discount'      :[10, 8, 15, 10]}

df1=pd.DataFrame(newDict)
df1

newDfIndex = df1.set_index(['Date','Product'])
newDfIndex
```

```python
df1.set_index(["Date","Product","Price"],inplace=True)
df1
```

```python
df1.set_index(["Discount"],inplace=True)
df1
```
