**<i>Simple Filtering</i>**<br>
**Column Filtering**<br>
Syntax: df[['column1', 'column2']]

Example: df[['FirstName', 'LastName']]

**Row Filtering (Slicing by Position)**<br>
Syntax: df[startRow:endRow]

Example: df[2:4]

**Row & Column Filtering**<br>
.loc (Label-Based Indexing)

Syntax: df.loc[rowLabels, ['ColumnNames']]

Example: df.loc[2:4, ['FirstName', 'LastName']]

.iloc (Integer-Position Indexing)

Syntax: df.iloc[rowIndex, columnIndexNumbers]

Example: df.iloc[2:4, 1:3]

**<i>Conditional Filtering</i>**<br>
<i>Create the condition:</i><br>
cond01 = df["Age"] >= 42<br>
cond02 = df['Airline'] == "Jet Airways"<br>

- Insert into Indexing and Slicing:
df[cond01 & cond02]
- Insert into .loc:
df.loc[cond01 & cond02, [ColumnNames]]
  - Example: df.loc[cond01 & cond02, ["Firstname", "LastName"]]
- Insert into .iloc:
df.iloc[cond01 & cond02, [ColumnIndexNumbers]]
  - Example: df.iloc[cond01 & cond02, 1:2]

```python
file_Path = "E:\Ethans Tech\Data\Small Airline.csv"
df = pd.read_csv(file_Path)
df.head()
```

```python
# df.loc[col1 | col2 | col3, [ColumnList]]

cond1 = df['Age'] >= 42
cond2 = df['Airline'] == 'Jet Airways'
cond3 = df['Full Name'].str.contains('Gurjar')

# Return full Dataframe with conditions
df.loc[cond1 & cond2 & cond3]
```

```python
# df.loc[col1 | col2 | col3, [ColumnList]]

cond1 = df['Age'] >= 42
cond2 = df['Airline'] == 'Jet Airways'
cond3 = df['Full Name'].str.contains('Gurjar')

# Use of loc to filter rows based on multiple conditions and select specific columns
df.loc[cond1 & cond2 & cond3, ['Age', 'Airline', 'Full Name']]
```
