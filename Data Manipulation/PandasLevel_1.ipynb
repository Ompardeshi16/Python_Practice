<i>Reading and Saving DataFrame</i>
1. **Data Input:**
- read_csv() : Read a CSV file into a DataFrame.
- read_excel() : Read an Excel file into a DataFrame.
- read_json() : Read a JSON file into a DataFrame.
- read_sql() : Read from a SQL database into a DataFrame.
2. **Save File:**
- to_csv() : Write a DataFrame to a CSV file.
- to_excel() : Write a DataFrame to an Excel file.
- to_json() : Write a DataFrame to a JSON file.
- to_sql() : Write records stored in a DataFrame to a SQL database.

**Data Exploration**
- df.info() — Displays summary of DataFrame including index, columns, non-null values, and data types.
- df.describe() — Generates descriptive statistics for numeric columns (count, mean, std, min, max, etc.).
- df.columns — Lists all column names of the DataFrame.
- df['columnName'].unique() — Returns all unique values in a specific column.
- df.value_counts() — Returns counts of unique rows or values in a Series.

Mean : Average Value <br>
Median: Middle Value <br>

**<i>Simple Filtering</i>**<br>
**Column Filtering**<br>
Syntax: df[['column1', 'column2']]

Example: df[['FirstName', 'LastName']]

**Row Filtering (Slicing by Position)**<br>
Syntax: df[startRow:endRow]

Example: df[2:4]

**Row & Column Filtering**<br>
.loc (Label-Based Indexing)

Syntax: df.loc[rowLabels, ['ColumnNames']]

Example: df.loc[2:4, ['FirstName', 'LastName']]

.iloc (Integer-Position Indexing)

Syntax: df.iloc[rowIndex, columnIndexNumbers]

Example: df.iloc[2:4, 1:3]

```python
# Column Filter
FilterOne = ['City']
df['City']
```

```python
# Row Filtering
df[0:4]
```

```python
# Row And Column Filtering
# Syntax df.loc[RowIndexNumber,[ColumnNames]]
ColumnList = ['City','Airline']
df.loc[0:4,ColumnList]
```

```python
# Row And Column Filtering
# Syntax df.iloc[RowIndexNumber,ColumnIndexNumber]
df.iloc[0:4,0:3]
```
