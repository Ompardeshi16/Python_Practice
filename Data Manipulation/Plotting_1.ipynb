# <h1 style= color:red> Ploting</h1>
<i>Matplotlib</i>

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
filepath = "D:\Data Science Classes Ethan Tech\Dataset\Airline1.csv"
df = pd.read_csv(filepath)
df.head()
```

- Bar Plot

```python
plt.figure(figsize=(8,3)) # Set the figure size for better visibility
plt.bar(df['Airline'], df['Price'])
plt.xticks(rotation=30) # X-axis labels rotated for better visibility

font = 15
plt.xlabel('Airline', fontsize=font)
plt.ylabel('Price', fontsize=font)
plt.title('Airline vs Price', fontsize=font)

plt.show()
```

<i>Seaborn</i>

```python
import seaborn as sns

sns.barplot(x=df['Airline'], y=df['Price'])
plt.xticks(rotation=40) # X-axis labels rotated for better visibility
plt.show()
```

Task: Create a line plot, count plot, and violin plot using the same dataset.


```python
df.head(2)
```

```python
# Line plot
plt.figure(figsize=(15,3))
# plt.plot(df['Airline'], df['Price'])
sns.lineplot(data=df, x='Airline', y='Price', marker='o', markersize=8, color='orange')
plt.xticks(rotation=90)

font = 15
plt.xlabel('Airline', fontsize=font)
plt.ylabel('Price', fontsize=font)
plt.title("Line Plot", fontsize=font)
plt.show()
```

```python
# Count plot
plt.figure(figsize=(8,4))
ax = sns.countplot(data=df, x='Airline')
ax.bar_label(ax.containers[0], fmt='%d', label_type='edge', fontsize=8) # Add count labels on top of the bars
plt.xticks(rotation=90)

font = 15
plt.xlabel('Airline', fontsize=font)
plt.ylabel('Count', fontsize=font)
plt.title("Count Plot", fontsize=font)
plt.show()
```

From Countplot we can see that Indigo has the highest number of flights and SpiceJet has the lowest number of flights and the rest of the airlines have almost equal number of flights.

```python
# Count plot
plt.figure(figsize=(8,4))
ax = sns.countplot(data=df, x='Destination')
ax.bar_label(ax.containers[0], fmt='%d', label_type='edge', fontsize=8)
'''
ax.containers[0] is used to access the bars in the count plot, 
and bar_label is used to add labels on top of the bars. 
The fmt='%d' specifies that the labels should be formatted as integers, 
and label_type='edge' places the labels at the edge of the bars. The fontsize=8 sets the font size of the labels.
'''
plt.xticks(rotation=30)

font = 15
plt.xlabel('Destination', fontsize=font)
plt.ylabel('Count', fontsize=font)
plt.title("Count Plot", fontsize=font)
plt.show()
```

```python
# Violin plot
plt.figure(figsize=(8,5))
sns.violinplot(data=df, y='Airline', hue='Destination')
# plt.xticks(rotation=90)

font = 15
# plt.xlabel('Price', fontsize=font)
plt.ylabel('Airline', fontsize=font)
plt.title("Violin Plot")
plt.show()
```

```python
# Violin plot
plt.figure(figsize=(12,5))
sns.violinplot(data=df, y='Price', hue='Airline')
# plt.xticks(rotation=90)

font = 15
# plt.xlabel('Price', fontsize=font)
plt.ylabel('Price', fontsize=font)
plt.title("Violin Plot")
plt.show()
```

```python
# Boxplot
plt.figure(figsize=(15,5))
sns.boxplot(data=df, y='Price', hue='Airline')
plt.title("Box Plot")
plt.show()
```

```python
# Boxplot
plt.figure(figsize=(12,5))
sns.boxplot(data=df, y='Price', hue='Destination')
plt.xticks(rotation=90)
plt.title("Box Plot")
plt.show()
```
