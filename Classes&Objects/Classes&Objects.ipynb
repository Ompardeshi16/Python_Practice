- Argument use by Function

```python
class test:

    def compare(self,i):
        if i<10:
            print("Less than 10")
        else:
            print("Greter than 10")

    def print(self,i):
        for i in range(1,i):
            print(i)


obj1 = test()
obj1.compare(5)
obj1.print(5)
```

- Argument use by Class(Constructor -> __init__(self) )

```python
class test:
    def __init__(self,i):
        self.i = i

    def compare(self):
        if self.i<10:
            print("Less than 10")
        else:
            print("Greter than 10")

    def print(self):
        for i in range(1,self.i):
            print(i)


obj1 = test(3)
obj1.compare()
obj1.print()
```
